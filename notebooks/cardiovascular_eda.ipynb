# Cardiovascular dataset EDA (sulianova/cardiovascular-disease-dataset)
**Why this dataset is NOT used to train a lipid model:** its `cholesterol` column is an ordinal category (1 normal, 2 above normal, 3 well above normal) and the target `cardio` is presence of cardiovascular disease. Neither is a measured lipid value nor a labelled lipid abnormality, so the lipid module stays rule-based.

Download with `scripts/download_datasets.sh`; this notebook only runs if the file is present.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.width', 140)
p='../datasets/cardio/cardio_train.csv'
if os.path.exists(p):
    d=pd.read_csv(p,sep=';'); print(d.shape); display(d.head()); display(d['cholesterol'].value_counts()); display(pd.crosstab(d['cholesterol'],d['cardio'],normalize='index'))
else:
    print('Dataset not downloaded – see datasets/README.md')
```
